```python
import pandas as pd

df = pd.read_csv("insider-trades.csv")
```

```python
df.info()
```

```python

buys = df[(df["transaction"] == "buy") & (df["value_over_2b_check_filing"] != "yes")]


print(buys)
```

```python
import sqlite3

connection = sqlite3.connect("dataForTrade.db")

buys.to_sql("insider_buys", connection, if_exists="replace", index=False)

theSQLTest = pd.read_sql("SELECT * FROM insider_buys WHERE insider = 'HELMETAG CARL'", connection)

print(theSQLTest)

connection.close()
```
